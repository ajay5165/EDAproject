
9. How Often Does Recorded Speed Exceed the Speed Limit?

```python
overspeeding_cases = df[df['Recorded_Speed'] > df['Speed_Limit']]
overspeeding_count = overspeeding_cases.shape[0]
total_violations = df.shape[0]
percentage_overspeeding = (overspeeding_count / total_violations) * 100
percentage_overspeeding
```

11. What are the most common violations for each vehicle type?

12. What's the average fine amount by gender of the driver?

13. Are towed vehicles more likely to have unpaid fines?

```python
df[df['Towed'] == 'Yes']['Fine_Paid'].value_counts(normalize=True) * 100
```

14. Are there more violations during day or night?

```python
df['Hour'] = pd.to_datetime(df['Time'], errors='coerce').dt.hour
df['Time_Of_Day'] = df['Hour'].apply(lambda x: 'Day' if 6 <= x <= 18 else 'Night')
df['Time_Of_Day'].value_counts()
```

15.How have traffic violations trended over time on a monthly basis

```python
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

# Group by month and count violations
monthly_violations = df['Date'].dt.to_period('M').value_counts().sort_index()

# Plot
plt.figure(figsize=(6, 4))
monthly_violations.plot(kind='bar', color='brown')
plt.title('Monthly Trend of Traffic Violations')
plt.xlabel('Month')
plt.ylabel('Number of Violations')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

16. Which vehicle types are involved in the most violations?

```python
df['Vehicle_Type'].value_counts().plot(kind='pie', autopct='%1.1f%%', startangle=140)
plt.title("Violations by Vehicle Type")
plt.ylabel("")
plt.show()
```
